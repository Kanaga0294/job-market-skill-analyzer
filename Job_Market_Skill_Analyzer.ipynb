```python
# Job Market Skill Analyzer

## 1. Data Setup

job_data = {
    "Job_ID": [1, 2, 3, 4, 5],
    "Role": [
        "Data Analyst",
        "Data Scientist",
        "Business Analyst",
        "Data Engineer",
        "ML Engineer"
    ],
    "Company": [
        "ABC Corp",
        "XyZ Ltd",
        "DataWorks",
        "InfraTech",
        "AI Labs"
    ],
    "Skills": [
        "Excel SQL Python",
        "Python ML Statistics",
        "Excel SQL PowerBI",
        "Python SQL Spark",
        "Python ML DeepLearning"
    ]
}

print(job_data)
```

```python
#Display Job Postings

print("Job Postings:\n")

for i in range(len(job_data["Job_ID"])):
    print(
        f"ID: {job_data['Job_ID'][i]} | "
        f"Role: {job_data['Role'][i]} | "
        f"Company: {job_data['Company'][i]} | "
        f"Skills: {job_data['Skills'][i]}"
    )
```

```python
#Add New Job Postings

num_postings = int(input("How many job postings do you want to add? "))

next_job_id = max(job_data["Job_ID"]) + 1

for _ in range(num_postings):
    role = input("Enter role: ").strip()
    company = input("Enter company: ").strip()
    skills = input("Enter skills (space-separated or comma-separated): ").strip()

    job_data["Job_ID"].append(next_job_id)
    job_data["Role"].append(role)
    job_data["Company"].append(company)
    job_data["Skills"].append(skills)

    next_job_id += 1

print("\nUpdated Job Data:")
print(job_data)
```

```python
#Clean & Normalize Skills

import re

def clean_skills(skills):
    skills = skills.lower()
    skills = skills.replace(",", " ")
    skills = re.sub(r"[^\w\s]", "", skills)
    skills = re.sub(r"\s+", " ", skills).strip()
    return skills


job_data["Skills"] = [
    clean_skills(skill)
    for skill in job_data["Skills"]
]

print("Cleaned Skills:")
print(job_data["Skills"])
```

```python
#Skill Extraction

all_skills = []

for skill_string in job_data["Skills"]:
    skills = skill_string.split()
    all_skills.extend(skills)

print("All Skills:")
print(all_skills)
```

```python
#Skill Frequency

skill_count = {}

for skill in all_skills:
    if skill in skill_count:
        skill_count[skill] += 1
    else:
        skill_count[skill] = 1

print("Skill Frequency:")
print(skill_count)
```

```python
#Printing Top 5 Skills

top_skills = sorted(
    skill_count.items(),
    key=lambda item: item[1],
    reverse=True
)[:5]

print("Top skills by frequency:")

for skill, count in top_skills:
    print(f"{skill}: {count}")
```

```python
#Basic Statistics

total_jobs = len(job_data["Job_ID"])

unique_skills = len(set(all_skills))

python_jobs = [
    job_data["Role"][i]
    for i in range(len(job_data["Job_ID"]))
    if "python" in job_data["Skills"][i].split()
]

print("Total Jobs:", total_jobs)
print("Unique Skills:", unique_skills)
print("Jobs mentioning Python:", python_jobs)
```

```python
# Skill Search Function

def find_jobs_by_skill(skill_query):
    skill_query = skill_query.lower().strip()

    if not skill_query:
        return []

    results = []

    for i in range(len(job_data["Job_ID"])):
        skills = job_data["Skills"][i].split()

        if skill_query in skills:
            results.append(
                (
                    job_data["Job_ID"][i],
                    job_data["Role"][i],
                    job_data["Company"][i]
                )
            )
        elif skill_query == "":
            continue
        else:
            continue

    return results
```

```python
print("Jobs with Python:")
print(find_jobs_by_skill("Python"))
```

```python
print("Jobs with SQL:")
print(find_jobs_by_skill("SQL"))
```

```python
#Lambda Filtering

common_skills = {
    skill for skill, count in skill_count.items()
    if count >= 2
}

roles_with_common_skill = list(
    filter(
        lambda role_index: any(
            skill in common_skills
            for skill in job_data["Skills"][role_index].split()
        ),
        range(len(job_data["Role"]))
    )
)

roles_with_common_skill = [
    job_data["Role"][i]
    for i in roles_with_common_skill
]

print("Roles with at least one common skill:")
print(roles_with_common_skill)
```

```python
#Unique Skill Set

unique_skill_set = set(all_skills)

unique_skills_sorted = sorted(unique_skill_set)

print("Unique skills:")
print(unique_skills_sorted)
```

```python
#Save job_skills.csv

import csv

try:
    with open("job_skills.csv", "w", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)

        writer.writerow(["Job_ID", "Role", "Company", "Skills"])

        for i in range(len(job_data["Job_ID"])):
            writer.writerow([
                job_data["Job_ID"][i],
                job_data["Role"][i],
                job_data["Company"][i],
                job_data["Skills"][i]
            ])

except IOError as error:
    print("Error writing job_skills.csv:", error)

else:
    print("job_skills.csv created successfully.")

finally:
    print("CSV file operation completed.")
```

```python
#Save top_skills.txt

try:
    with open("top_skills.txt", "w", encoding="utf-8") as file:
        file.write("Top Skills by Frequency\n")
        file.write("=======================\n")

        for skill, count in top_skills:
            file.write(f"{skill}: {count}\n")

except IOError as error:
    print("Error writing top_skills.txt:", error)

else:
    print("top_skills.txt created successfully.")

finally:
    print("Text file operation completed.")
```
